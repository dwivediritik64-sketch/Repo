```python

# AI INNOVATIVE ASSESSMENT - PART II
# Name: Ritik Dwivedi
# Roll No: 54
# Group 4: WhatsApp Chat

import re
import nltk
import pandas as pd

from nltk.stem import PorterStemmer, WordNetLemmatizer
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from IPython.display import display

nltk.download('wordnet')
nltk.download('omw-1.4')

# Original WhatsApp message
text = """Hi everyone! Our project presentation is scheduled for tomorrow at 10:00 AM in Lab 3. Please bring your laptops, presentation slides, and printed reports. Don't forget to revise your assigned topics before the meeting. If anyone has any doubts, please message me tonight so we can discuss them together. See you all tomorrow!"""

# Tokenization and basic cleaning
words = re.findall(r'\b[a-zA-Z]+\b', text.lower())

# Remove common stop words
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS

clean_words = [word for word in words if word not in ENGLISH_STOP_WORDS]

print("AI INNOVATIVE ASSESSMENT - PART II")
print("Roll No: 54 | Group: 4")
print("Text Type: WhatsApp Chat")
print("\nOriginal Text:")
print(text)
print("\nTotal Words:", len(words))
print("Words After Stop Word Removal:", len(clean_words))
```

```python

# Apply Stemming

ps = PorterStemmer()

stemmed_words = [ps.stem(word) for word in clean_words]

print("=" * 50)
print("1. STEMMING OUTPUT")
print("=" * 50)

result = pd.DataFrame({
    "Original Word": clean_words,
    "Stemmed Word": stemmed_words
})

display(result)
print("Total Stemmed Words:", len(stemmed_words))
```

```python

# Apply Lemmatization

lemmatizer = WordNetLemmatizer()

lemmatized_words = [
    lemmatizer.lemmatize(word) for word in clean_words
]

print("=" * 50)
print("2. LEMMATIZATION OUTPUT")
print("=" * 50)

result = pd.DataFrame({
    "Original Word": clean_words,
    "Lemmatized Word": lemmatized_words
})

display(result)
print("Total Lemmatized Words:", len(lemmatized_words))
```

```python

# Generate Bag of Words Representation

cv = CountVectorizer()
bow_matrix = cv.fit_transform([" ".join(clean_words)])

bow_df = pd.DataFrame({
    "Word": cv.get_feature_names_out(),
    "Frequency": bow_matrix.toarray()[0]
})

bow_df = bow_df.sort_values(
    by="Frequency", ascending=False
).reset_index(drop=True)

print("=" * 50)
print("3. BAG OF WORDS (BoW) OUTPUT")
print("=" * 50)

display(bow_df)
print("Total Unique Words:", len(bow_df))
print("Total Word Frequency:", bow_df["Frequency"].sum())
```

```python

# Generate TF-IDF Representation

tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform([" ".join(clean_words)])

tfidf_df = pd.DataFrame({
    "Word": tfidf.get_feature_names_out(),
    "TF-IDF Value": tfidf_matrix.toarray()[0]
})

tfidf_df = tfidf_df.sort_values(
    by="TF-IDF Value", ascending=False
).reset_index(drop=True)

print("=" * 50)
print("4. TF-IDF REPRESENTATION")
print("=" * 50)

display(tfidf_df.round(4))

print("Total Features:", len(tfidf_df))
print("Highest TF-IDF Value:",
      round(tfidf_df["TF-IDF Value"].max(), 4))
```
